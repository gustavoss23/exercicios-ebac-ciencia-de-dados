### Quais são os hyperparâmetros do RF? Pra que serve cada um deles?

- **n_estimators**: Número de árvores na floresta. Quanto maior, melhor a performance (até certo ponto), mas com maior custo computacional.

- **criterion**: Função usada para medir a qualidade das divisões.  
  - Classificação: `"gini"` ou `"entropy"`  
  - Regressão: `"squared_error"` (ou `"mse"`), `"absolute_error"` (ou `"mae"`)

- **max_depth**: Profundidade máxima das árvores. Limitar ajuda a evitar overfitting.

- **min_samples_split**: Mínimo de amostras necessário para dividir um nó interno. Valores maiores reduzem o overfitting.

- **min_samples_leaf**: Mínimo de amostras em um nó folha. Evita folhas com poucas amostras, o que também reduz overfitting.

- **max_features**: Número de características consideradas em cada divisão. Pode ser `"sqrt"`, `"log2"`, um número inteiro ou uma fração.

- **max_leaf_nodes**: Número máximo de nós folha por árvore. Limita o crescimento e a complexidade da árvore.

- **min_impurity_decrease**: Divisões só ocorrem se a redução da impureza for maior que esse valor.

- **bootstrap**: Define se as amostras para cada árvore são obtidas com reposição. `True` ajuda na diversidade das árvores.

- **oob_score**: Se `True`, calcula o desempenho do modelo com amostras fora da bolsa (out-of-bag).

- **n_jobs**: Número de núcleos usados para paralelizar o treinamento. `-1` usa todos os disponíveis.

- **random_state**: Semente usada para garantir resultados reproduzíveis.

- **verbose**: Nível de mensagens exibidas durante o treinamento. Útil para monitorar o progresso.

- **warm_start**: Se `True`, adiciona mais árvores ao modelo já treinado (sem reiniciar do zero).

- **class_weight**: Ajusta o peso das classes em problemas de classificação. Útil para dados desbalanceados.

- **ccp_alpha**: Parâmetro de poda de custo-complexidade. Remove ramos irrelevantes e simplifica o modelo.

- **max_samples**: Número ou fração de amostras usadas por árvore (se `bootstrap=True`). Ajuda a acelerar e diversificar as árvores.
