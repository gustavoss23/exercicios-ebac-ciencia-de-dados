## 1- Calcule de forma numérica a integral da função $x^3$ avaliada entre os pontos $1$ e $2$. Identifique o número de retângulos que você usou.  

```python
import numpy as np

def f(x):
    return x**3

a = 1
b = 2


n1 = 4
delta_x1 = (b - a) / n1
x_vals1 = np.linspace(a, b - delta_x1, n1)
f_vals1 = f(x_vals1)
area_4 = np.sum(f_vals1) * delta_x1

n2 = 100
delta_x2 = (b - a) / n2
x_vals2 = np.linspace(a, b - delta_x2, n2)
f_vals2 = f(x_vals2)
area_100 = np.sum(f_vals2) * delta_x2


print(f"Aproximação numérica com {n1} retângulos: {area_4}")
print(f"Aproximação numérica com {n2} retângulos: {area_100}")
```
