    4. CART (Árvore de decisão)

Os métodos baseados no uso de derivada são B. Gradiente descendente e C. Newton-Raphson.

2. Dada uma base de dados com uma variável resposta $y$ e um conjunto de variáveis explicativas. Considere uma estrutura de um modelo de regressão. Explique com suas palavras por que não é possível obter parâmetros que forneçam um erro quadrático médio (EQM) menor que o obtido com estimadores de mínimos quadrados.

O método dos mínimos quadrados busca ajustar um modelo de regressão encontrando os parâmetros que minimizam a soma dos quadrados dos erros. Isso significa que, por definição, ele encontra os melhores parâmetros possíveis para minimizar exatamente o erro quadrático médio (EQM). Como o método já entrega os parâmetros que minimizam o EQM, não é possível obter outros parâmetros que resultem em um EQM menor. Qualquer outro conjunto de parâmetros levaria a um EQM igual ou maior.
