## 1. Cite 5 diferenças entre o Random Forest e o AdaBoost

- **Estratégia de Ensemble**: O Random Forest usa Bagging (Bootstrap Aggregation). O AdaBoost usa Boosting.

- **Treinamento dos Modelos**: O Random Forest treina as árvores independentemente e em paralelo. O AdaBoost treina os classificadores sequencialmente, cada um corrigindo os erros do anterior.

- **Combinação dos Modelos**: O Random Forest usa votação majoritária (classificação) ou média (regressão). O AdaBoost combina os modelos com pesos baseados na performance de cada um.

- **Sensibilidade a Outliers**: O Random Forest é menos sensível a ruído e outliers. O AdaBoost é mais sensível, pois enfatiza os erros nas próximas iterações.

- **Paralelização**: O Random Forest é fácil de paralelizar devido à independência dos estimadores. O AdaBoost é difícil de paralelizar, pois cada etapa depende da anterior.

## 2. Acesse o link Scikit-learn– adaboost , leia a explicação (traduza se for preciso) e crie um jupyter notebook contendo o exemplo do AdaBoost.

```python
from sklearn.model_selection import cross_val_score
from sklearn.datasets import load_iris
from sklearn.ensemble import AdaBoostClassifier

X, y = load_iris(return_X_y=True)
clf = AdaBoostClassifier(n_estimators=100, algorithm="SAMME",)
scores = cross_val_score(clf, X, y, cv=5)
scores.mean()
```

## 3. Cite 5 Hyperparametros importantes no AdaBoost.

- **n_estimators**: Número de classificadores fracos a serem usados no ensemble.

- **learning_rate**: Controla o peso de cada classificador fraco. Baixos valores exigem mais estimadores.

- **estimator**: O modelo fraco utilizado (ex: uma árvore de decisão rasa).

- **algorithm**: Algoritmo de atualização dos pesos. Pode ser "SAMME" ou "SAMME.R" (padrão).

- **random_state**: Controla a aleatoriedade para reprodutibilidade dos resultados.

## 4. Utilize o GridSearch para encontrar os melhores hyperparametros para o conjunto de dados do exemplo (load_iris)

```python
from sklearn.datasets import load_iris
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.metrics import accuracy_score

X, y = load_iris(return_X_y=True)

clf = AdaBoostClassifier(algorithm="SAMME")

param_grid = {
    'n_estimators': [50, 100, 150, 200],
    'learning_rate': [0.001, 0.01, 0.1, 1],
    'random_state': [0, 10, 20]
}

grid_search = GridSearchCV(
    estimator=clf,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)

grid_search.fit(X, y)

best_params = grid_search.best_params_
best_score = grid_search.best_score_

print("Melhores hiperparâmetros encontrados:")
print(best_params)
print("Melhor acurácia encontrada:")
print(best_score)
```
