### 1. Monte um passo a passo para o Bagging

- Gerar subconjuntos aleatórios com reposição (Bootstrap)

A partir do conjunto de dados original com N amostras, o bagging cria B subconjuntos (normalmente, B é o número de modelos a serem treinados). Cada subconjunto é gerado com reposição: isso significa que a mesma amostra pode aparecer mais de uma vez no mesmo subconjunto.

- Treinar modelos independentes

Um modelo (geralmente do mesmo tipo, como uma árvore de decisão) é treinado em cada subconjunto. Como cada subconjunto é diferente, os modelos aprendem de forma ligeiramente diferente.



- Combinar as previsões

Para regressão a média das previsões dos modelos é usada como a previsão final. Para classificação usa-se o voto da maioria (classe mais votada entre os modelos).

- Resultado final

O modelo final é mais robusto, tem menor variância e melhor desempenho em dados fora da amostra (dados de teste). Reduz o risco de overfitting, especialmente em modelos como árvores de decisão.



### 2. Explique com suas palavras o Bagging

O bagging (do inglês Bootstrap Aggregating) é uma técnica de ensemble usada para melhorar a estabilidade e a acurácia de modelos de machine learning, especialmente modelos instáveis como árvores de decisão. Ela ajuda a reduzir o overfitting e aumentar a robustez do modelo.

Exemplo prático (classificação com árvores de decisão):
1. Dados originais com 1000 amostras.

2. Gera-se 10 subconjuntos (bootstrap) de 1000 amostras cada.

3. Treina-se 10 árvores de decisão, uma por subconjunto.

4. Para um novo exemplo, as 10 árvores votam pela classe.

A classe mais votada é a predição final.

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import BaggingClassifier
from sklearn.metrics import accuracy_score

# 1. Gerar dataset com 1000 amostras e 20 features
X, y = make_classification(
    n_samples=1000,       # número de amostras
    n_features=20,        # número de features
    n_informative=15,     # features realmente úteis
    n_redundant=5,        # features redundantes
    n_classes=2,          # binário (0 ou 1)
    random_state=42
)

# 2. Dividir entre treino e teste
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 3. Criar o classificador base (árvore de decisão)
base_model = DecisionTreeClassifier()

# 4. Criar o BaggingClassifier com 10 árvores
bagging_model = BaggingClassifier(
    estimator=base_model,
    n_estimators=20,        # número de modelos
    bootstrap=True,         # amostragem com reposição
    random_state=42
)

# 5. Treinar o modelo
bagging_model.fit(X_train, y_train)

# 6. Fazer previsões
y_pred = bagging_model.predict(X_test)

# 7. Avaliar desempenho
acc = accuracy_score(y_test, y_pred)
print(f"Acurácia com Bagging (scikit-learn): {acc:.2f}")
```
